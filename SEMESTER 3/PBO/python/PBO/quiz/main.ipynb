Buat pilihan menu bidang datar yang dapat dipilih user, yaitu persegi panjang, bujur sangkar, atau lingkaran. Untuk setiap bangun datar yang dipilih, program akan menghitungkan luas dan keliling bangun tersebut. Nilai panjang, lebar, sisi, dan nilai jari-jari lingkaran diinputkan oleh user sesuai dengan bangun yang dipilihnya. Selama user masih ingin memilih, menu akan ditampilkan. Jika user sudah tidak mau, menu tidak akan ditampilkan.

```python
print("Selamat datang!!")
print("Masukkan Pilihan\n1. Persegi Panjang\n2. Bujur Sangkar\n3. Lingkaran\n4. Keluar")

while True:
    pilihan = int(input("Pilih: "))


    if pilihan == 1:
        print("Persegi Panjang")
        panjang1 = int(input("Panjang: "))
        lebar1 = int(input("Lebar: "))
        luas1 = panjang1 * lebar1
        keliling1 = 2 * (panjang1 + lebar1)
        print(f"Panjang nya adalah: {panjang1}, Lebar nya adalah: {lebar1}")
    elif pilihan == 2:
        print("Bujur Sangkar")
        sisi = int(input("Sisi: "))
        luas2 = sisi * sisi
        keliling2 = 4 * sisi
    elif pilihan == 3:
        print("Lingkaran")
        jari = int(input("Jari: "))
        luas3 = 3.14 * (jari * jari)
        keliling3 = 2 * 3.14 * jari
    elif pilihan == 4:
        break
```

```python
import math

# Fungsi untuk menampilkan dan memilih menu
def menu():
    print("Pilih bangun datar:")
    print("1. Persegi Panjang")
    print("2. Bujur Sangkar")
    print("3. Lingkaran")
    choice = input("Masukkan nomor pilihan: ")
    return choice

# Fungsi untuk menerima inputan panjang dan lebar persegi panjang
def input_persegi_panjang():
    panjang = float(input("Masukkan panjang persegi panjang: "))
    lebar = float(input("Masukkan lebar persegi panjang: "))
    return panjang, lebar

# Fungsi untuk menerima inputan sisi bujur sangkar
def input_bujur_sangkar():
    sisi = float(input("Masukkan sisi bujur sangkar: "))
    return sisi

# Fungsi untuk menerima inputan jari-jari lingkaran
def input_lingkaran():
    jari_jari = float(input("Masukkan jari-jari lingkaran: "))
    return jari_jari

# Fungsi untuk menghitung luas dan keliling persegi panjang
def hitung_persegi_panjang(panjang, lebar):
    luas = panjang * lebar
    keliling = 2 * (panjang + lebar)
    return luas, keliling

# Fungsi untuk menghitung luas dan keliling bujur sangkar
def hitung_bujur_sangkar(sisi):
    luas = sisi ** 2
    keliling = 4 * sisi
    return luas, keliling

# Fungsi untuk menghitung luas dan keliling lingkaran
def hitung_lingkaran(jari_jari):
    luas = math.pi * (jari_jari ** 2)
    keliling = 2 * math.pi * jari_jari
    return luas, keliling

# Main program
while True:
    choice = menu()

    if choice == '1':
        panjang, lebar = input_persegi_panjang()
        luas, keliling = hitung_persegi_panjang(panjang, lebar)
    elif choice == '2':
        sisi = input_bujur_sangkar()
        luas, keliling = hitung_bujur_sangkar(sisi)
    elif choice == '3':
        jari_jari = input_lingkaran()
        luas, keliling = hitung_lingkaran(jari_jari)
    else:
        print("Pilihan tidak valid.")
        continue

    print(f"Luas: {luas}")
    print(f"Keliling: {keliling}")

    ulangi = input("Lanjutkan (ya/tidak)? ")
    if ulangi.lower() != 'ya':
        break
```
