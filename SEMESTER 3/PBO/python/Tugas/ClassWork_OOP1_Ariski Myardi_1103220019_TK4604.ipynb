#### Soal 1
Buatlah kelas Segitiga yang memiliki atribut dan method yang sama dengan kelas Polygon tapi memiliki tambahan method untuk menentukan jenis segitiga dan method untuk menghitung keliling segitiga.

```python
class Poligon:
    def __init__(self, sisi):
        self.sisi = sisi

    def jumlah_sisi(self):
        return self.sisi

    def hitung_keliling(self):
        return sum(self.sisi)


class Segitiga(Poligon):
    def __init__(self, sisi):
        super().__init__(sisi)

    def jenis_segitiga(self):
        sisi_unik = len(set(self.sisi))

        if sisi_unik == 1:
            return "Segitiga Sama Sisi"
        elif sisi_unik == 2:
            return "Segitiga Sama Kaki"
        else:
            return "Segitiga Sembarang"

    def hitung_keliling_segitiga(self):
        return super().hitung_keliling()


# Contoh penggunaan:
segitiga_sama_sisi = Segitiga([5, 5, 5])
segitiga_sama_kaki = Segitiga([5, 5, 8])
segitiga_sembarang = Segitiga([3, 4, 5])

print("Jenis Segitiga:", segitiga_sama_sisi.jenis_segitiga())
print("Keliling Segitiga:", segitiga_sama_sisi.hitung_keliling_segitiga())

print("Jenis Segitiga:", segitiga_sama_kaki.jenis_segitiga())
print("Keliling Segitiga:", segitiga_sama_kaki.hitung_keliling_segitiga())

print("Jenis Segitiga:", segitiga_sembarang.jenis_segitiga())
print("Keliling Segitiga:", segitiga_sembarang.hitung_keliling_segitiga())
```

Buatlah kelas Square dan kelas Rectangle yang mewarisi kelas Polygon dengan kekhasan perilaku masing-masing, yaitu dapat menghitung luas dan keliling. Selain itu, saat user menginputkan panjang sisi, jika tidak memenuhi syarat sebagai square dan rectangle maka program akan mengingatkan user untuk menginputkan panjang yang memenuhi kriteria.

```python
class Poligon:
    def __init__(self, sisi):
        self.sisi = sisi

    def jumlah_sisi(self):
        return self.sisi

    def hitung_keliling(self):
        return sum(self.sisi)

    def hitung_luas(self):
        pass  # Metode ini akan diimplementasikan oleh kelas turunan


class Persegi(Poligon):
    def __init__(self, sisi):
        if self.apakah_persegi_valid(sisi):
            super().__init__([sisi, sisi, sisi, sisi])
        else:
            print("Panjang sisi tidak memenuhi kriteria sebagai persegi")

    def apakah_persegi_valid(self, sisi):
        return sisi > 0

    def hitung_luas(self):
        return self.sisi[0] ** 2


class PersegiPanjang(Poligon):
    def __init__(self, panjang, lebar):
        if self.apakah_persegi_panjang_valid(panjang, lebar):
            super().__init__([panjang, lebar, panjang, lebar])
        else:
            print("Panjang dan lebar tidak memenuhi kriteria sebagai persegi panjang")

    def apakah_persegi_panjang_valid(self, panjang, lebar):
        return panjang > 0 and lebar > 0

    def hitung_luas(self):
        return self.sisi[0] * self.sisi[1]


# Contoh penggunaan:
panjang_sisi_persegi = float(input("Masukkan panjang sisi persegi: "))
persegi = Persegi(panjang_sisi_persegi)
if persegi.jumlah_sisi():
    print("Keliling Persegi:", persegi.hitung_keliling())
    print("Luas Persegi:", persegi.hitung_luas())

panjang_persegi_panjang = float(input("Masukkan panjang persegi panjang: "))
lebar_persegi_panjang = float(input("Masukkan lebar persegi panjang: "))
persegi_panjang = PersegiPanjang(panjang_persegi_panjang, lebar_persegi_panjang)
if persegi_panjang.jumlah_sisi():
    print("Keliling Persegi Panjang:", persegi_panjang.hitung_keliling())
    print("Luas Persegi Panjang:", persegi_panjang.hitung_luas())
```

#### Soal 3
Buatlah sebuah program  yang mengimplementasikan inheritance yang terdiri atas:
1. Class Pegawai yang memiliki properti nama, nip, dan divisi serta memiliki method konstruktor untuk mengisi property nama dan nip, dan method untuk menginputkan data divisi oleh user
2. Class PegawaiTetap, yang memiliki properti dan method persis sama dengan yang ada pada Class Pegawai, namun ada tambahan properti jabatan  dan method untuk menginputkan data jabatan oleh user

```python
class Pegawai:
    def __init__(self, nama, nip):
        self.nama = nama
        self.nip = nip
        self.divisi = ""

    def input_divisi(self):
        self.divisi = input("Masukkan divisi pegawai: ")

    def tampilkan_informasi(self):
        print(f"Nama: {self.nama}")
        print(f"NIP: {self.nip}")
        print(f"Divisi: {self.divisi}")


class PegawaiTetap(Pegawai):
    def __init__(self, nama, nip):
        super().__init__(nama, nip)
        self.jabatan = ""

    def input_jabatan(self):
        self.jabatan = input("Masukkan jabatan pegawai tetap: ")

    def tampilkan_informasi(self):
        super().tampilkan_informasi()
        print(f"Jabatan: {self.jabatan}")


# Contoh penggunaan:
pegawai1 = Pegawai("Ariski Myardi", "12345")
pegawai1.input_divisi()
print("\nInformasi Pegawai:")
pegawai1.tampilkan_informasi()

pegawai2 = PegawaiTetap("Cahyo Ilahi", "67890")
pegawai2.input_divisi()
pegawai2.input_jabatan()
print("\nInformasi Pegawai Tetap:")
pegawai2.tampilkan_informasi()
```

#### Soal 4
Buatlah program oop yang berisi class Mahasiswa (yang pekan lalu dibuat) dan kelas turunannya class MahasiswaKelas yang dapat menyimpan data kelas berupa kode kelas, dosen wali, jumlah mahasiswa, dan data mahasiswa mencakup nama, NIM, nilai makul 1, nilai makul 2, nilai makul 3. Kelas turunan ini juga dapat menghitung nilai rata-rata kelas untuk mata kuliah 1, 2, dan 3.

```python
class Mahasiswa:
    def __init__(self, nama, nim):
        self.nama = nama
        self.nim = nim

class MahasiswaKelas(Mahasiswa):
    def __init__(self, nama, nim, kode_kelas, dosen_wali, jumlah_mahasiswa, nilai_makul):
        super().__init__(nama, nim)
        self.kode_kelas = kode_kelas
        self.dosen_wali = dosen_wali
        self.jumlah_mahasiswa = jumlah_mahasiswa
        self.nilai_makul = nilai_makul

    def hitung_rata_rata(self, nilai_makul):
        return sum(nilai_makul) / len(nilai_makul)

# Contoh penggunaan
mahasiswa1 = MahasiswaKelas("Ariski Myardi", "1103220019", "TK-46-04", "Pak Faisal", 30, [90, 97, 92])
print("Rata-rata nilai Mahasiswa 1:", mahasiswa1.hitung_rata_rata(mahasiswa1.nilai_makul))
```
