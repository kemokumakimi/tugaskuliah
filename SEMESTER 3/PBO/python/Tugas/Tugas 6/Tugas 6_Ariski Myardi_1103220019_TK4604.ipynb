Buatlah program OOP yang dapat menyimpan data mahasiswa beserta data nilainya. Data mahasiswa beserta nilainya diinputkan oleh user. Program juga dapat menghitung nilai akhir (NA) berupa nilai rata-rata dari seluruh nilai dan mengkonversinya ke dalam indeks berikut ini.
<br>Indeks A , nilai>80 
<br>Indeks B , 60<= nilai <=80 
<br>Indeks C , 50<= nilai <60 
<br>Indeks D , 40<=nilai <50 
<br>Indeks E , nilai<40
<br><br>Program terdiri atas 2 class yaitu class Mahasiswa dan class Nilai. 
<br>Class Nilai memiliki atribut nama MK, nilai 1, nilai 2, nilai 3, nilai akhir, dan nilai indeks. 
<br>Class Mahasiswa dapat menyimpan data mahasiswa berupa nama, NIM, kelas, nilai MK PBO dan AI. Nilai MK PBO dan AI disimpan dalam bentuk obyek bertipe kelas Nilai. 
<br><br>Running program untuk menyimpan data 3 mahasiswa dan tampilkan ke layar data seluruh mahasiswa yang diinputkan dalam format (dipisahkan oleh tab):
<br>No.	Nama	NIM	 	Kelas	Nama MK		NA``Indeks	

```python
class Nilai:
    def __init__(self, nama_mk, nilai1, nilai2, nilai3):
        self.nama_mk = nama_mk
        self.nilai1 = nilai1
        self.nilai2 = nilai2
        self.nilai3 = nilai3
        self.nilai_akhir = self.hitung_nilai_akhir()
        self.indeks = self.konversi_ke_indeks()

    def hitung_nilai_akhir(self):
        return (self.nilai1 + self.nilai2 + self.nilai3) / 3

    def konversi_ke_indeks(self):
        nilai_akhir = self.hitung_nilai_akhir()
        if nilai_akhir > 80:
            return 'A'
        elif 60 <= nilai_akhir <= 80:
            return 'B'
        elif 50 <= nilai_akhir < 60:
            return 'C'
        elif 40 <= nilai_akhir < 50:
            return 'D'
        else:
            return 'E'


class Mahasiswa:
    def __init__(self, nama, nim, kelas, nilai_pbo, nilai_ai):
        self.nama = nama
        self.nim = nim
        self.kelas = kelas
        self.nilai_pbo = nilai_pbo
        self.nilai_ai = nilai_ai

    def tampilkan_data(self, nomor):
        print(f"{nomor}\t{self.nama}\t{self.nim}\t{self.kelas}\t{self.nilai_pbo.nama_mk}\t\t{self.nilai_pbo.nilai_akhir:.2f}\t{self.nilai_pbo.indeks}")
        print(f"{nomor}\t{self.nama}\t{self.nim}\t{self.kelas}\t{self.nilai_ai.nama_mk}\t\t{self.nilai_ai.nilai_akhir:.2f}\t{self.nilai_ai.indeks}")


mahasiswa1 = Mahasiswa("Kimi", "1100019", "04", Nilai("PBO", 85, 78, 92), Nilai("AI", 75, 88, 80))
mahasiswa2 = Mahasiswa("adsf", "1104123", "02", Nilai("PBO", 70, 65, 75), Nilai("AI", 88, 92, 78))
mahasiswa3 = Mahasiswa("adfadf", "1102301", "03", Nilai("PBO", 60, 55, 70), Nilai("AI", 45, 50, 55))

print("No.\tNama\tNIM\tKelas\tNama MK\t\tNA\tIndeks")
mahasiswa1.tampilkan_data(1)
mahasiswa2.tampilkan_data(2)
mahasiswa3.tampilkan_data(3)
```
