#### Soal 2
Buatlah program untuk mencari akar-akar persamaan kuadrat (𝑎𝑥^2+𝑏𝑥+𝑐=0) dengan nilai a, b, dan c berupa bilangan bulat yang diinputkan oleh user. 

```python

a = float(input("Masukkan nilai a: "))
b = float(input("Masukkan nilai b: "))
c = float(input("Masukkan nilai c: "))

diskriminan = b**2 - 4*a*c

if diskriminan > 0:
    x1 = (-b + (diskriminan ** 0.5)) / (2*a)
    x2 = (-b - (diskriminan ** 0.5)) / (2*a)
    print(f"Akar-akar persamaan kuadrat adalah x1 = {x1} dan x2 = {x2}")
elif diskriminan == 0:
    x1 = -b / (2*a)
    print(f"Akar ganda persamaan kuadrat adalah x1 = x2 = {x1}")
else:
    real_part = -b / (2*a)
    imaginary_part = (abs(diskriminan) ** 0.5) / (2*a)
    print(f"Akar-akar kompleks persamaan kuadrat adalah:")
    print(f"x1 = {real_part} + {imaginary_part}i")
    print(f"x2 = {real_part} - {imaginary_part}i")
```

#### Soal 3
Ada beberapa data buah, dengan komponen : nama, kualitas dan harga.
<br>Jenis buah ada 3: jeruk dengan harga 1000, mangga 2000, dan apel 3000, masing-masing untuk kualitas 2. 
<br>Jika jeruk kualitas 1, maka harganya menjadi 1,5 kali lipat, dan kualitas 3 menjadi 75% dari harga kualitas 2. 
<br>Mangga ada 2 kualitas, 1 dan 2 dengan harga kualitas 1 adalah 2 kali lipat dari kualitas 2.  Sedangkan apel hanya ada kualitas 2. 
<br>Masukan/input program adalah nilai uang, nama buah, dan kualitas. 
<br>Luaran/output program adalah nama buah, kualitas dan harga. 
<br>Misal jika inputannya adalah sbb: untuk masukan ’5000’ (untuk uang), jeruk (nama), 3 (untuk kualitas), maka akan didapatkan keluaran :’Anda membeli jeruk kualitas 3 dengan harga 750, dan mendapatkan 6 buah jeruk dengan sisa uang 500’. 

```python
def hitung_harga(uang, namaBuah, kualitas):
    if namaBuah == "jeruk":
        harga = 1000
    elif namaBuah == "mangga":
        if kualitas == 1:
            harga = 2000 * 2
        else:
            harga = 2000
    elif namaBuah == "apel":
        harga == 3000

    if kualitas == 1:
        harga *= 1.5
    elif kualitas == 3:
        harga * 0.75

    jumlahBuah = uang // harga
    sisaUang = uang % harga

    print(f"Anda membeli {namaBuah} kualitas {kualitas} dengan harga {int(harga)}, dan mendapatkan {jumlahBuah} buah {namaBuah} dengan sisa uang {sisaUang} ")



uang  = int(input("Masukkan Jumlah Uang"))
namaBuah = input("Masukkan Nama Buah")
kualitas = int(input("Masukkan Kualitas Buah"))

hitung_harga(uang,namaBuah,kualitas)
```

#### Soal 4
Buatlah program tarif parkir dengan ketentuan :
<br>Biaya 2 jam pertama adalah Rp6000
<br>Jika lebih dari 2 jam, maka biayanya ditambah dengan Rp2500 setiap jam
<br>Program menerima inputan dari user berupa lama waktu parkir (dalam jam)

```python
jam = int(input("berapa jam anda telah parkir?"))
duaJam = 6000
if jam == 0:
    print("tidak ada")
elif jam <= 2:
    print(f"kamu telah parkir selama {jam}, dan harganya adalah {duaJam}")
elif jam >= 2:
    biayaParkir= ((jam - 2) * 2500) + 6000
    print(f"kamu telah parkir selama {jam}, dan harganya adalah {biayaParkir}")
```

#### Soal 5
Buatlah program untuk menentukan kelompok usia seseorang yang diinput oleh pengguna, dengan ketentuan sebagai berikut :
<br>a) Jika usia 5 tahun ke bawah (usia<=5), maka dia adalah balita.
<br>b) Jika usia di atas 5 sampai 12 tahun , maka dia adalah anak-anak.
<br>c) Jika usia di atas 12 tahun dan usia di bawah 18 tahun maka dia adalah remaja.
<br>d) Jika usia di atas 18 tahun, maka dia adalah dewasa.

```python
umur = int(input("masukkan umur:"))

if umur <= 5:
    print("dia adalah balita")
elif umur <= 12:
    print("dia adalah anak anak")
elif umur <= 18:
    print("dia adalah remaja")
elif umur >= 18:
    print("dia adalah dewasa")
else: print("tidak valid")
```
