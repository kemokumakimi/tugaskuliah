### Soal 1
Modifikasilah program yang dikerjakan di Quiz 1 untuk menampilkan list bilangan ganjil, genap, dan bilangan prima dari range tertentu menjadi program yang menggunakan 4 buah fungsi berikut ini.
<br>1. fungsi untuk menampilkan dan menerima pilihan menu. Termasuk di dalamnya menerima inputan batas bawah dan batas atas dari user.
<br>2. fungsi untuk menampilkan bilangan ganjil sesuai dengan batas bawah dan batas atas yang diinputkan user
<br>3. fungsi untuk menampilkan bilangan genap sesuai dengan batas bawah dan batas atas yang diinputkan user
<br>4. fungsi untuk menampilkan bilangan prima sesuai dengan batas bawah dan batas atas yang diinputkan user


```python
def primasolve(n):
    if n <= 1:
            return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def ganjil(batas_bawah, batas_atas):
    ganjil = [num for num in range(
        batas_bawah, batas_atas + 1) if num % 2 != 0]
    print("Bilangan ganjil dalam range tersebut adalah:", ganjil)


def genap(batas_bawah, batas_atas):
    genap = [num for num in range(batas_bawah, batas_atas + 1) if num % 2 == 0]
    print("Bilangan genap dalam range tersebut adalah:", genap)


def prima(batas_bawah, batas_atas):

    prima = [num for num in range(
        batas_bawah, batas_atas + 1) if primasolve(num)]
    print("Bilangan prima dalam range tersebut adalah:", prima)


batas_bawah = int(input("Masukkan batas bawah range: "))
batas_atas = int(input("Masukkan batas atas range: "))

while True:
    print("Menu Pilihan:")
    print("1. Menampilkan bilangan ganjil")
    print("2. Menampilkaan bilangan genap")
    print("3. Menampilkan bilangan prima")
    print("4. Keluar")

    pilihan = input("Pilih opsi (1/2/3/4): ")

    if pilihan == '1':
        ganjil(batas_bawah, batas_atas)
    elif pilihan == '2':
        genap(batas_bawah, batas_atas)
    elif pilihan == '3':
        prima(batas_bawah, batas_atas)
    elif pilihan == '4':
        print("Terima kasih telah menggunakan program ini.")
        break
    else:
        print("Opsi yang Anda pilih tidak valid. Silakan pilih kembali.")

```

### Soal 2
Buatlah program yang dapat:
<br>1. menerima sejumlah data awal mahasiswa berupa Nama, NIM, Nilai UTS, dan Nilai UAS dari user. Jumlah data yang akan dimasukkan diinputkan oleh user.
<br>2. menuliskan data mahasiswa tersebut ke dalam file .txt
<br>3. menambahkan data mahasiswa tambahan kemudian menuliskan tambahan tersebut ke file
<br>3. setiap selesai menjalankan menu tertentu, program akan menanyakan kepada user apakah user masih ingin memilih menu yang lain atau tidak. Selama user masih ingin memilih menu maka menu akan selalu ditampilkan kembali

```python
def menerima_data():
    mahasiswa = []
    n = int(input("Berapa banyak mahasiswa? "))

    for i in range(n):
        print(f"Masukkan data ke-{i+1}")
        nama = input("Nama: ")
        nim = input("Nim: ")
        prodi = input("Prodi: ")
        uts = int(input("UTS: "))
        uas = int(input("UAS: "))
        data_mahasiswa    ={
            "Nama": nama,
            "Nim": nim,
            "Prodi": prodi,
            "UTS": uts,
            "UAS": uas
        }
        mahasiswa.append(data_mahasiswa)

    return mahasiswa

def menambahkan_data(mahasiswa):
    print("Menambahkan data Mahasiswa: ")
    nama = input("Nama: ")
    nim = input("Nim: ")
    prodi = input("Prodi: ")
    uts = int(input("UTS: "))
    uas = int(input("UAS: "))
    data_mahasiswa ={
        "Nama": nama,
        "Nim": nim,
        "Prodi": prodi,
        "UTS": uts,
        "UAS": uas
        }

    mahasiswa.append(data_mahasiswa)

def input_data(mahasiswa):
    with open("file.txt", "a") as file:
        for data in mahasiswa:
            file.write(f"Nama: {data['Nama']}\nNim: {data['Nim']}\nProdi: {data['Prodi']}\nUTS: {data['UTS']}, UAS: {data['UAS']} \n")
            
mahasiswa = []

while True:
    print("Selamat di Program..")
    print("1. Menerima Data Mahasiswa")
    print("2. Menambahkan Data Mahasiswa")
    print("3. Input Data Mahasiswa ke File")
    print("4. Keluar?")

    pilihan = int(input("Apa yang ingin anda lakukan? "))

    if pilihan == 1:
        mahasiswa = menerima_data()
    elif pilihan == 2:
        menambahkan_data(mahasiswa)
    elif pilihan == 3:
        input_data(mahasiswa)
    elif pilihan == 4:
        print("Terima Kasih telah menggunakan Program ini!!")
        break
```
