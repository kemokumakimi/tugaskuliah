#### Soal
Buatlah program yang menerima inputan 3 angka. Dari 3 angka tersebut, tentukan angka yang terbesar. Cetak juga di layar urutan ketiga nilai tersebut dari yang terbesar ke yang paling kecil.

```python
bil1 = int(input("Masukkan angka 1: "))
bil2 = int(input("Masukkan ankga 2: "))
bil3 = int(input("Masukkan angka 3: "))

if bil1 > bil2 and bil1 > bil3:
    print("Bilangan 1 angka paling besar")
elif bil2 > bil1 and bil2 > bil3:
    print("Bilangan 2 angka paling besar")
elif bil3 > bil1 and bil3 > bil2:
    print("Bilangan 3 angka paling besar")

print("Kamu adalah orang")
daftar_angka = [bil1, bil2, bil3]
print("Daftar Angka: ",daftar_angka)

daftar_angka.sort(reverse=True)

print("Urutan dari besar ke kecil: ",daftar_angka)
```

#### Soal
Buatlah program untuk menampilkan list bilangan ganjil, genap, dan bilangan prima dari range tertentu. Batas bawah range dan batas atas range diinputkan oleh user.
<br>Buatlah tampilan menu pilihan yang dapat dipilih user: menampilkan list bilangan ganjil atau bilangan genap, atau bilangan prima. Selama user masih ingin memilih, menu akan ditampilkan. Jika user sudah tidak mau, menu tidak akan ditampilkan.

```python
def prima(n):
        if n <= 1:
            return False
        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False
        return True


while True:
        print("Menu Pilihan:")
        print("1. Menampilkan bilangan ganjil")
        print("2. Menampilkan bilangan genap")
        print("3. Menampilkan bilangan prima")
        print("4. Keluar")
        
        pilihan = input("Pilih opsi (1/2/3/4): ")

        if pilihan == '1':
            batas_bawah = int(input("Masukkan batas bawah range: "))
            batas_atas = int(input("Masukkan batas atas range: "))
            ganjil = [num for num in range(batas_bawah, batas_atas + 1) if num % 2 != 0]
            print("Bilangan ganjil dalam range tersebut adalah:", ganjil)
        elif pilihan == '2':
            batas_bawah = int(input("Masukkan batas bawah range: "))
            batas_atas = int(input("Masukkan batas atas range: "))
            genap = [num for num in range(batas_bawah, batas_atas + 1) if num % 2 == 0]
            print("Bilangan genap dalam range tersebut adalah:", genap)
        elif pilihan == '3':
            batas_bawah = int(input("Masukkan batas bawah range: "))
            batas_atas = int(input("Masukkan batas atas range: "))
            prima = [num for num in range(batas_bawah, batas_atas + 1) if prima(num)]
            print("Bilangan prima dalam range tersebut adalah:", prima)
        elif pilihan == '4':
            print("Terima kasih telah menggunakan program ini.")
            break
        else:
            print("Opsi yang Anda pilih tidak valid. Silakan pilih kembali.")


```

Tampilkan 5 angka random dari 1-100 tapi harus berkelipatan 5 :
<br>Jika hasilnya <= 60 maka tampilkan tulisan "Kurang" 
<br>Jika >60 dan <= 70 maka tampilkan tulisan "Cukup" 
<br>Jika >70 dan <= 80 maka tampilkan tulisan "Baik" 
<br>Jika >80 maka tampilkan tulisan "Luar Biasa" 
<br>Contoh :
<br>65 Cukup
<br>25 Kurang
<br>85 Luar biasa dst.

<br>Notes: gunakan kode berikut ini untuk menghasilkan bilangan random:
<br>import random
<br>random.randint(batasBawahRange,batasAtasRange)

```python
import random

for i in range(5):
    angka_random = random.randint(1, 20) * 5

    if angka_random <= 60:
        print(f"{angka_random} Kurang")
    elif angka_random <= 70:
        print(f"{angka_random} Cukup")
    elif angka_random <= 80:
        print(f"{angka_random} Baik")
    else:
        print(f"{angka_random} Luar Biasa")
```
