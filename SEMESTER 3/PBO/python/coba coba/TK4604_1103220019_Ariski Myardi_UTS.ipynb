#### Soal 1
Buatlah program yang dapat menerima data mahasiswa yang diinputkan user berupa Nama, NIM, nilai UTS, dan nilai UAS. Data mahasiswa tersebut disimpan dalam bentuk dictionary yang mengandung list. Program dapat menghitungkan nilai akhir sekaligus memberikan indeks nilai dalam bentuk huruf.
<br><br>Program terdiri atas beberapa fungsi program yaitu:
<br>	fungsi main() sebagai program utama
<br>	fungsi hitungNA(uts,uas) untuk menghitung Nilai Akhir dengan bobot 40% untuk nilai UTS dan 60% untuk nilai UAS
<br>	fungsi indeksHuruf(nilaiAkhir) untuk menentukan indeks nilai dalam bentuk huruf dengan ketentuan sbb.
<br>&emsp;A : 80 < nilai <= 100
<br>&emsp;B: 65 < nilai <= 80
<br>&emsp;C: 50 < nilai <= 65
<br>&emsp;D: 40 < nilai <= 50
<br>&emsp;E: nilai < 40
<br>	fungsi meanKelas(listNA) untuk menghitung nilai rata-rata kelas dari listNa yang berisi nilai akhir semua mahasiswa yang ada di data mahasiswa
<br>	fungsi tampilData(dataMhs,rataKelas) untuk menampilkan keseluruhan data akhir dengan format sbb (menggunakan \t untuk tab):
<br>No. Nama &emsp;NIM &emsp;UTS &emsp;UAS &emsp;NA &emsp;Indeks
<br>…
<br>…
<br>Nilai rata-rata kelas adalah …



```python


def hitungNA(uts, uas):
    return 0.4 * uts + 0.6 * uas

def indeksHuruf(nilaiAkhir):
    if nilaiAkhir > 80:
        return 'A'
    elif nilaiAkhir > 65:
        return 'B'
    elif nilaiAkhir > 50:
        return 'C'
    elif nilaiAkhir > 40:
        return 'D'
    else:
        return 'E'

def meanKelas(listNA):
    return sum(listNA) / len(listNA)

def tampilData(dataMhs, rataKelas):
    print("No.\tNama\tNIM\tUTS\tUAS\tNA\tIndeks")
    for i, mahasiswa in enumerate(dataMhs, start=1):
        nama, nim, uts, uas = mahasiswa['nama'], mahasiswa['nim'], mahasiswa['uts'], mahasiswa['uas']
        nilaiAkhir = hitungNA(uts, uas)
        indeks = indeksHuruf(nilaiAkhir)
        print(f"{i}\t{nama}\t{nim}\t{uts}\t{uas}\t{nilaiAkhir}\t{indeks}")
    
    print(f"Nilai rata-rata kelas adalah {rataKelas:.2f}")


dataMhs = []
jumlahMahasiswa = int(input("Masukkan jumlah mahasiswa: "))
    
for i in range(jumlahMahasiswa):
    nama = input(f"Masukkan nama mahasiswa ke-{i + 1}: ")
    nim = input(f"Masukkan NIM mahasiswa ke-{i + 1}: ")
    uts = float(input(f"Masukkan nilai UTS mahasiswa ke-{i + 1}: "))
    uas = float(input(f"Masukkan nilai UAS mahasiswa ke-{i + 1}: "))
    
    mahasiswa = {'nama': nama, 'nim': nim, 'uts': uts, 'uas': uas}
    dataMhs.append(mahasiswa)
    
    listNA = [hitungNA(mhs['uts'], mhs['uas']) for mhs in dataMhs]
    rataKelas = meanKelas(listNA)
    
    tampilData(dataMhs, rataKelas)
```

#### Soal 2
Buatlah program konversi yang terdiri atas dua pilihan menu yaitu Konversi Hari dan Koversi Detik. 
<br><br>Pada menu Konversi Hari, program menerima inputan user berupa jumlah hari yang akan dikoversi menjadi sejumlah tahun, bulan, minggu dan hari. Perhitungan menggunakan asumsi 1 tahun = 365 hari dan 1 bulan = 30 hari. Contoh sbb:
<br>Jumlah hari yang akan dikonversi = 990 hari
<br>Hasil konversi: 990 hari = 2 tahun 8 bulan 2 minggu 6 hari
<br><br>Pada menu Konversi Detik, program menerima inputan user berupa jumlah detik yang akan dikoversi menjadi sejumlah jam, menit, dan detik. Contoh sbb:
<br>Jumlah detik yang akan dikonversi = 10000 detik
<br>Hasil konversi: 1000 detik = 2 jam 46 menit 40 detik
<br><br>Program akan terus menampilkan pilihan menu selama user masih ingin melakukan konversi.


```python
def konversi_hari_ke_tahun_bulan_minggu_hari(hari):
    tahun = hari // 365
    sisa_hari = hari % 365
    bulan = sisa_hari // 30
    sisa_hari %= 30
    minggu = sisa_hari // 7
    sisa_hari %= 7

    return tahun, bulan, minggu, sisa_hari

def konversi_detik_ke_jam_menit_detik(detik):
    jam = detik // 3600
    sisa_detik = detik % 3600
    menit = sisa_detik // 60
    sisa_detik %= 60

    return jam, menit, sisa_detik

while True:
    print("Menu Konversi:")
    print("1. Konversi Hari")
    print("2. Konversi Detik")
    print("3. Keluar")

    pilihan = input("")

    if pilihan == '1':
        jumlah_hari = int(input("Masukkan jumlah hari yang akan dikonversi: "))
        tahun, bulan, minggu, sisa_hari = konversi_hari_ke_tahun_bulan_minggu_hari(jumlah_hari)
        print(f"Hasil konversi: {jumlah_hari} hari = {tahun} tahun {bulan} bulan {minggu} minggu {sisa_hari} hari")
    elif pilihan == '2':
        jumlah_detik = int(input("Masukkan jumlah detik yang akan dikonversi: "))
        jam, menit, sisa_detik = konversi_detik_ke_jam_menit_detik(jumlah_detik)
        print(f"Hasil konversi: {jumlah_detik} detik = {jam} jam {menit} menit {sisa_detik} detik")
    elif pilihan == '3':
        print("Terima kasih, program selesai.")
        break
    
```

#### Soal 3
Sebuah toko online memiliki data barang, kode barang, jumlah stok, dan harga jual yang disimpan di dalam dictionary.
<br>Buatlah program yang memudahkan pembeli untuk membeli barang yang diinginkan. Program menampilkan semua pilihan barang dan user dapat memilih nomor barang yang akan dibeli dan menginputkan jumlah yang akan dibelinya. Program akan menghitungkan total pembeliannya dan mengupdate jumlah stok barang.
<br>Selama user masih akan membeli barang, program akan menampilkan daftar barang yang dapat dibelinya.
<br>Setelah user selesai membeli barang, program akan menuliskan rekap barang yang dibeli berisi nama barang, kode barang, jumlah yang dibeli serta total pembayaran ke dalam file dengan nama “transaksi.txt”.
<br>Program juga akan mencatatkan update daftar barang beserta stoknya ke file "update barang.txt"
<br>Program kemudian membaca kedua file tersebut dan menampilkan rekapannya ke layar. 


```python

daftarBarang = {'nama': ['sabun', 'pasta gigi', 'shampoo', 'sikat gigi', 'kapas', 'indomie', 'biskuit Oreo', 'pulpen', 'buku', 'penghapus'],
                'stok': [20, 18, 20, 15, 10, 25, 30, 10, 8, 12],'harga jual': [3000, 2500, 4000, 2500, 2000, 3500, 4000, 2500, 4000, 2000]}

transaksi = {'nama Barang': [], 'jumlah pembelian': [], 'total harga': []}

print("Berikut ini daftar barang, jumlah stok beserta harganya.")
print("No. \tNama Barang \tJumlah Stok \t Harga Jual")
for i in range(10):
    print(f"{1+i}. \t{daftarBarang['nama'][i]}\t{daftarBarang['stok'][i]}\t{daftarBarang['harga jual'][i]}")

totalBayar = 0
belanja = input("Apakah akan membeli barang (Ya/Tidak)? ")

while belanja.lower() == "ya":
    for i in range(10):
        if daftarBarang['stok'][i] == 0:
            print(f"{daftarBarang['nama'][i]} sudah habis")
    
    noBarang = int(input("Pilih nomor barang yang akan dibeli: "))
    jumlahBeli = int(input("Jumlah barang yang akan dibeli: "))
    
    if jumlahBeli <= daftarBarang['stok'][noBarang]:

        totalHarga = daftarBarang['harga jual'][noBarang] * jumlahBeli
        
        transaksi['nama Barang'].append(daftarBarang['nama'][noBarang])
        transaksi['jumlah pembelian'].append(jumlahBeli)
        transaksi['total harga'].append(totalHarga)
        
        daftarBarang['stok'][noBarang] -= jumlahBeli
        
        belanja = input("Masih akan membeli barang? (Ya/Tidak) ")
    else:
        print("Stok barang tidak mencukupi.")

with open("transaksi.txt", "w") as file:
    for i in range(len(transaksi['nama Barang'])):
        file.write(f"Nama Barang: {transaksi['nama Barang'][i]}, Jumlah Pembelian: {transaksi['jumlah pembelian'][i]}, Total Harga: {transaksi['total harga'][i]}\n")

with open("update_barang.txt", "w") as file:
    for i in range(10):
        file.write(f"Nama Barang: {daftarBarang['nama'][i]}, Stok Tersedia: {daftarBarang['stok'][i]}\n")

with open("transaksi.txt", "r") as file:
    print("Rekap Transaksi:")
    for line in file:
        print(line.strip())

with open("update_barang.txt", "r") as file:
    print("Update Stok Barang:")
    for line in file:
        print(line.strip())
```
