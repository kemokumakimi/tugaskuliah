Buatlah program yang terdiri atas 2 kelas, yaitu class Pasien dan class Aplikasi. 
1. class Pasien berisi atribut yang dimiliki oleh pasien yaitu nama (public), NIK (public), nama dokter (protected), dan penyakit yang diidap (privat). 
2. class Aplikasi berisi method untuk memberi pilihan menu kepada admin, menambah data pasien dan menampilkan data pasien. Menu yang ditawarkan hanya 2: menambah data pasien atau menampilkan data pasien.
Tambahkan atribut dan method lain jika diperlukan.

```python
class Pasien:
    def __init__(self, nama, NIK, nama_dokter, penyakit):
        self.nama = nama
        self.NIK = NIK
        self._nama_dokter = nama_dokter 
        self.__penyakit = penyakit   

    def get_nama_dokter(self):
        return self._nama_dokter

    def get_penyakit(self):
        return self.__penyakit


class Aplikasi:
    def __init__(self):
        self.daftar_pasien = []

    def tambah_pasien(self):
        nama = input("Masukkan nama pasien: ")
        NIK = input("Masukkan NIK pasien: ")
        nama_dokter = input("Masukkan nama dokter: ")
        penyakit = input("Masukkan penyakit yang diidap: ")

        pasien_baru = Pasien(nama, NIK, nama_dokter, penyakit)
        self.daftar_pasien.append(pasien_baru)
        print("Data pasien berhasil ditambahkan.")

    def tampilkan_pasien(self):
        if not self.daftar_pasien:
            print("Belum ada data pasien.")
        else:
            print("Daftar Pasien:")
            for pasien in self.daftar_pasien:
                print(f"Nama: {pasien.nama}, NIK: {pasien.NIK}, Dokter: {pasien.get_nama_dokter()}, Penyakit: {pasien.get_penyakit()}")

```

```python
aplikasi = Aplikasi()

while True:
        print("\nMenu Aplikasi:")
        print("1. Tambah Data Pasien")
        print("2. Tampilkan Data Pasien")
        print("0. Keluar")

        pilihan = input("Masukkan pilihan (0/1/2): ")

        if pilihan == "1":
            aplikasi.tambah_pasien()
        elif pilihan == "2":
            aplikasi.tampilkan_pasien()
        elif pilihan == "0":
            print("Terima Kasih telah menggunakan program")
            break
        else:
            print("Pilihan tidak valid.")
```
