Masing-masing operasi menggunakan dua inputan bilangan saja. Hasil operasinya ditampilkan di aplikasi GUInya. Feel free untuk merancang tampilan GUI dan layoutnya.

```python
from tkinter import *

win1 = Tk()
win1.geometry("750x500")

label0 = Label(win1, text="Input form", font=('Arial 20')).pack(side=TOP, pady=15)

label1 = Label(win1, text="Nilai A:", font=('Arial 12'))
T01 = Entry()
label1.place(x=100, y=100)
T01.place(x=180, y=100)

label2 = Label(win1, text="Nilai B:", font=('Arial 12'))
Td1 = Entry()
label2.place(x=350, y=100)
Td1.place(x=430, y=100)

result_label = Label(win1, text="Result = ", font=('Arial 20'))
result_label.pack(side=BOTTOM, pady=150)

def tambah():
    A = float(T01.get())
    B = float(Td1.get())
    result = A + B
    result_label.config(text="Result = " + str(result))


def kurang():
    A = float(T01.get())
    B = float(Td1.get())
    result =A - B
    result_label.config(text="Result = " + str(result))

def kali():
    A = float(T01.get())
    B = float(Td1.get())
    result = A * B
    result_label.config(text="Result = " + str(result))
    
def bagi():
    A = float(T01.get())
    B = float(Td1.get())
    result = A / B
    result_label.config(text="Result = " + str(result))


btn = Button(win1, text="+", font=('Arial 12'), command=tambah)
btn.place(x=300, y=350)

btn2 = Button(win1, text="-", font=('Arial 12'), command=kurang)
btn2.place(x=350, y=350)

btn3 = Button(win1, text="*", font=('Arial 12'), command=kali)
btn3.place(x=400, y=350)

btn4 = Button(win1, text="/", font=('Arial 12'), command=bagi)
btn4.place(x=450, y=350)
# btnhapus = Button(win1, text="C", font=("Arial 12"), command=hapus)
# btnhapus.place(x=500, y=350)

win1.bind('<Return>', lambda event: penjumlahan())
win1.mainloop()
```

### Program 4
Modifikasilah Program 2 di atas untuk menghasilkan program GUI yang dapat menghitung luas dan keliling persegi panjang. Hasil perhitungan luas dan keliling ditampilkan di aplikasi GUI. Feel free untuk merancang tampilan GUI dan layoutnya.

```python
from tkinter import *

win1 = Tk()
win1.geometry("750x500")

label0 = Label(win1, text="Input form", font=('Arial 20')).pack(side=TOP, pady=15)

label1 = Label(win1, text="Nilai A:", font=('Arial 12'))
T01 = Entry()
label1.place(x=100, y=100)
T01.place(x=180, y=100)

label2 = Label(win1, text="Nilai B:", font=('Arial 12'))
Td1 = Entry()
label2.place(x=350, y=100)
Td1.place(x=430, y=100)

result_label = Label(win1, text="Result = ", font=('Arial 20'))
result_label.pack(side=BOTTOM, pady=150)

def hasilLuas():
    A = float(T01.get())
    B = float(Td1.get())
    result = A * B
    result_label.config(text="Result = " + str(result))


def hasilKeliling():
    A = float(T01.get())
    B = float(Td1.get())
    result = 2 * (A + B)
    result_label.config(text="Result = " + str(result))




btn = Button(win1, text="Luas", font=('Arial 12'), command=hasilLuas)
btn.place(x=300, y=350)

btn2 = Button(win1, text="Keliling", font=('Arial 12'), command=hasilKeliling)
btn2.place(x=360, y=350)

# btnhapus = Button(win1, text="C", font=("Arial 12"), command=hapus)
# btnhapus.place(x=500, y=350)

win1.bind('<Return>', lambda event: penjumlahan())
win1.mainloop()
```
